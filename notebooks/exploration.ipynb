# Understanding Airbnb

Airbnb is an online platform that allows individuals and businesses to list properties or spaces for short-term stays. Listings can range from private rooms to entire homes or apartments.

The Airbnb marketplace connects hosts and guests and allows guests to compare listings based on factors such as:

- Location
- Price
- Room type
- Minimum stay requirements
- Reviews
- Availability

From a data-analysis perspective, Airbnb provides an interesting dataset because listing prices can vary substantially depending on location, accommodation type, demand-related factors, and host characteristics.

In this project, Airbnb NYC 2019 data is analyzed to understand these patterns and investigate which variables are associated with listing prices.

# Literature Review

Previous research on Airbnb pricing has shown that listing characteristics, location, accommodation type, reviews, host characteristics, and availability can be associated with differences in listing prices.

Research on Airbnb markets has commonly examined the role of location and property characteristics in explaining price variation. Studies have also investigated how host characteristics, reviews, and other listing attributes contribute to the pricing structure of short-term rental markets.

The literature provides a useful framework for this project because it suggests that Airbnb price is not determined by a single variable. Instead, multiple listing, host, location, and market-related characteristics can contribute to price differences.

This project applies exploratory data analysis to the NYC Airbnb 2019 dataset to investigate whether similar patterns can be observed in this particular dataset.

### 1. Location matters

### 3. Reviews provide useful information

### 4. Host characteristics can influence the marketplace

### 5. Airbnb pricing is multidimensional

# Outlier Analysis

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
price_outliers = df[
    (df["price"] < lower_bound) |
    (df["price"] > upper_bound)
]

print("Number of price outliers:", len(price_outliers))
print(
    "Percentage of price outliers:",
    round(len(price_outliers) / len(df) * 100, 2),
    "%"
)
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["price"]
)

plt.title("Price Outlier Analysis")
plt.xlabel("Price ($)")

plt.show()
```
